**17. Oblicz sumę ludności wszystkich Państw na rok 2010. 
W sytuacji gdy w danym kraju nie przeprowadzono
badania w roku 2010 wykorzystaj najnowsze z badań wcześniejszych.**

````python
from pyspark.sql.window import Window
# tu wprowadź swoje rozwiazanie
exploded_df = countriesDF.select(
    col("name"),
    explode(col("population")).alias("pop_struct")
).select(
    col("name"),
    col("pop_struct.year").alias("year"),
    col("pop_struct.value").alias("value")
)

# Krok 2: Filtrowanie danych
# Interesują nas tylko wpisy z roku 2010 lub wcześniejsze
filtered_df = exploded_df.filter(col("year") <= 2005)

# Krok 3: Użycie Window function do znalezienia "najnowszego z wcześniejszych"
# Grupujemy po nazwie kraju (partitionBy), sortujemy lata malejąco (orderBy desc),
# aby rok 2010 (lub najbliższy mu z dołu) był pierwszy.
windowSpec = Window.partitionBy("name").orderBy(desc("year"))

# Nadajemy numer wiersza (row_number). Wiersz z numerem 1 to ten najbardziej aktualny dla danego kraju.
ranked_df = filtered_df.withColumn("rank", row_number().over(windowSpec))

# Krok 4: Wybieramy tylko te wiersze, które są najnowszymi dostępnymi danymi (rank == 1)
latest_census_df = ranked_df.filter(col("rank") == 1)

# Krok 5: Sumujemy populację
result = latest_census_df.agg(sum("value").alias("total_population"))

# Wyświetlenie wyniku
result.show()
````

**18. Było ciężko? Nie wierzę.**

**Teraz już będzie z górki. Podaj nazwy i gęstość zaludnienia trzech krajów o największej gęstości zaludnienia w roku 2010.**

````python
exploded_df = countriesDF.select(
    col("name"),
    col("area"),
    explode(col("population")).alias("pop_struct")
).select(
    col("name"),
    col("area"),
    col("pop_struct.year").alias("year"),
    col("pop_struct.value").alias("population_value")
)

# Krok 2: Filtrowanie danych (rok <= 2010)
filtered_df = exploded_df.filter(col("year") <= 2010)

# Krok 3: Wybór najnowszego pomiaru dla każdego kraju (Window Function)
windowSpec = Window.partitionBy("name").orderBy(desc("year"))
latest_census_df = filtered_df.withColumn("rank", row_number().over(windowSpec)) \
    .filter(col("rank") == 1)

# Krok 4: Obliczenie gęstości zaludnienia (Density = Population / Area)
# Upewniamy się, że nie dzielimy przez zero (area != 0) - choć w tym zbiorze dane są zazwyczaj poprawne.
density_df = latest_census_df.filter(col("area") > 0) \
    .withColumn("density", col("population_value") / col("area"))

# Krok 5: Sortowanie malejąco po gęstości i ograniczenie do 5 wyników
top5_density = density_df.orderBy(desc("density")).limit(5)

# Wyświetlenie wyniku - ostatni wiersz w tabeli to poszukiwany 5. kraj
top5_density.show()
````


**19. Podaj trzy kraje o największym procencie ludności żyjącym w miastach powyżej 50 000 mieszkańców w roku
2010.**

````python
# Krok 1: Filtrowanie miast
cities_filtered = citiesDF.filter(col("population") > 50000)

# Krok 2: Normalizacja kodów krajów i agregacja
cities_large_pop = cities_filtered.withColumn(
    "country_code", 
    trim(lower(col("country")))
).groupBy("country_code").agg(
    sum("population").alias("large_cities_population")
)

# Krok 3: Explodowanie i filtrowanie roku 2010
countries_2010 = countriesDF.withColumn(
    "population_data", 
    explode("population")
).filter(
    col("population_data.year") == 2010
).select(
    col("name").alias("country_name"),
    trim(lower(col("code"))).alias("country_code"),
    col("population_data.value").alias("total_population")
)

# Krok 4: Join używając kodów krajów
result = countries_2010.join(
    cities_large_pop,
    "country_code",
    "inner"
).select(
    "country_name",
    "total_population",
    "large_cities_population"
)

# Krok 5: Obliczenie procentu
result = result.withColumn(
    "percentage",
    (col("large_cities_population") / col("total_population") * 100)
).orderBy(
    col("percentage").desc()
).limit(4)

result.select(
    "country_name",
    col("percentage").alias("percentage_in_large_cities")
).show()
````
